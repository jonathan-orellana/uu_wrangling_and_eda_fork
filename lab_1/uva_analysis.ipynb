#### UVA Analysis

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
foreign_gifts = pd.read_csv("ForeignGifts_edu.csv", low_memory=False)
```

```python
foreign_gifts[
    foreign_gifts["Institution Name"].str.contains("Virginia", case=False, na=False)
]["Institution Name"].value_counts()
```

```python
uva_gifts = foreign_gifts[
    foreign_gifts["Institution Name"] == "University of Virginia"
]

print(uva_gifts.shape)
uva_gifts.head()
```

```python
uva_total_gifts = uva_gifts["Foreign Gift Amount"].sum()

print(f"UVA total: ${uva_total_gifts:,.2f}")
```

```python
uva_average_gift = uva_gifts["Foreign Gift Amount"].mean()
uva_median_gift = uva_gifts["Foreign Gift Amount"].median()

print(f"UVA average gift: ${uva_average_gift:,.2f}")
print(f"UVA median gift: ${uva_median_gift:,.2f}")
```

```python
uva_country_totals = uva_gifts.groupby("Country of Giftor")["Foreign Gift Amount"].sum()

uva_country_totals.nlargest(3)
```

```python
uva_country_counts = uva_gifts["Country of Giftor"].value_counts()

uva_country_counts.nlargest(3)
```
