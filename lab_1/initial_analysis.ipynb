### Initial Analysis

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

##### A. Load the Foreign Gifts data.

```python
foreign_gifts = pd.read_csv("ForeignGifts_edu.csv", low_memory=False)
print(foreign_gifts.shape)
foreign_gifts.head()
```

#####  B. Describe the differences between different classes of gift types.

```python
foreign_gifts.columns
```

```python
foreign_gifts["Gift Type"].value_counts()
```

```python
pd.options.display.float_format = "{:,.2f}".format
gift_type_summary = foreign_gifts.groupby("Gift Type")["Foreign Gift Amount"].describe()
gift_type_summary
```



  The results show several notable differences between the different classes of gift types. Contracts are the most common with 17,274 records, followed by monetary gifts with 10,936. What caught my attention was the very low number of real estate gifts, with only 11 records. Despite having the smallest count, real estate has the largest mean at about $1.34 million. Contracts and monetary gifts have lower and somewhat similar means, with contracts averaging $625,794.51 and monetary gifts averaging $528,150.44.

  Another thing to point out is the large standard deviation across all three gift types, which shows that the gift amounts vary considerably. We can also see a large difference between the mean and median. Contracts have a median of $110,000, monetary gifts have a median of $50,000, and real estate gifts have a median of $667,555. Since the means are much higher than the medians, we can infer that some very large gift amounts are pulling the means upward.

  a. Which country gives the most money in total?

  Qatar is the country that gives the most money in total, with a total gift amount of $2.71 billion.

```python
country_total_gifts = foreign_gifts.groupby("Country of Giftor")["Foreign Gift Amount"].sum()
country_total_gifts
```

```python
country_total_gifts.nlargest(3)
```

b. Which country initiates the most gifts by count?

England is the country that initiates the most gifts, with a total of 3,655 gifts.

```python
country_gift_counts = foreign_gifts["Country of Giftor"].value_counts()
country_gift_counts.nlargest(3)
```

c. Which country gives the largest gifts on average?

Bermuda gives the largest gifts on average, with an average gift amount of approximately $7,688,837.37.

```python
country_average_gifts = foreign_gifts.groupby("Country of Giftor")["Foreign Gift Amount"].mean()
country_average_gifts.nlargest(3)
```

d. Which institution receives the most money in total?

Carnegie Mellon University receives the most money in total, with approximately $1.48 billion in foreign gifts.

```python
institution_total_gifts = foreign_gifts.groupby("Institution Name")["Foreign Gift Amount"].sum()
institution_total_gifts.nlargest(3)
```

e. Which institution receives the greatest number of gifts by count?

The University of California, Los Angeles receives the greatest number of gifts, with a total of 3,916 gift records


```python
institution_gift_counts = foreign_gifts["Institution Name"].value_counts()
institution_gift_counts.nlargest(3)
```

f. What is the average amount each university receives? Median amount? If they are different, why?

The mean is approximately $52.2 million, while the median is approximately $6.49 million. As we can see, the mean and median differ by a lot, suggesting that some very large values may be pulling the mean upward.

```python
average_institution_gifts = institution_total_gifts.mean()
median_institution_gifts = institution_total_gifts.median()

print(f"Average: ${average_institution_gifts:,.2f}")
print(f"Median: ${median_institution_gifts:,.2f}")
```

g. What is the largest flow of a single country to a single institution?

The largest flow from a single country to a single institution is from Qatar to Cornell University, with a total of approximately $1.02 billion.

```python
country_institution_total = foreign_gifts.groupby(
    ["Country of Giftor", "Institution Name"]
)["Foreign Gift Amount"].sum()

country_institution_total.nlargest(3)
```
